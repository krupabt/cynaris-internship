```python
import pandas as pd
import numpy as np

# Create a real-world style sales dataset
data = {
    "Customer": [" Anu", "Ravi ", "PRIYA", "anu", "Ravi", "Kiran ", "Priya", "ANU"],
    "Region": ["South", " north", "SOUTH ", "South", "NORTH", " west", "West", "south"],
    "Product": ["Laptop", "Phone", "Laptop", "Laptop", "Phone", "Tablet", "Laptop", "Laptop"],
    "Sales": [50000, 25000, np.nan, 52000, 26000, 18000, 60000, 250000],
    "Quantity": [2, 1, 3, 2, np.nan, 1, 3, 2],
    "Rating": [4.5, 4.0, np.nan, 4.2, 3.8, 4.1, 4.7, 5.0]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)

# BEFORE
before_rows = len(df)
before_columns = len(df.columns)
before_missing = int(df.isnull().sum().sum())
before_duplicates = int(df.duplicated().sum())

print("\nBEFORE CLEANING")
print("Rows:", before_rows)
print("Columns:", before_columns)
print("Missing values:", before_missing)
print("Duplicates:", before_duplicates)

# Remove duplicates
df = df.drop_duplicates()

# Standardise string values
for column in ["Customer", "Region", "Product"]:
    df[column] = df[column].str.strip().str.lower()

# Fill missing numeric values using median
for column in ["Sales", "Quantity", "Rating"]:
    df[column] = df[column].fillna(df[column].median())

# IQR outlier capping
numeric_columns = ["Sales", "Quantity", "Rating"]

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df[column] = df[column].clip(lower_bound, upper_bound)

# AFTER
after_rows = len(df)
after_columns = len(df.columns)
after_missing = int(df.isnull().sum().sum())
after_duplicates = int(df.duplicated().sum())

print("\nAFTER CLEANING")
print("Rows:", after_rows)
print("Columns:", after_columns)
print("Missing values:", after_missing)
print("Duplicates:", after_duplicates)

# Before / After comparison
quality_comparison = pd.DataFrame({
    "Metric": ["Rows", "Columns", "Missing Values", "Duplicate Rows"],
    "Before": [
        before_rows,
        before_columns,
        before_missing,
        before_duplicates
    ],
    "After": [
        after_rows,
        after_columns,
        after_missing,
        after_duplicates
    ]
})

print("\nDATA QUALITY COMPARISON")
display(quality_comparison)

# Summary statistics
print("\nSUMMARY STATISTICS")
display(df.describe())

# Save cleaned dataset
df.to_csv("cleaned_dataset.csv", index=False)

print("\ncleaned_dataset.csv created successfully!")
```
