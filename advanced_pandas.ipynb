```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)

print("Pandas version:", pd.__version__)
```

```python
customers = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105],
    "customer_name": ["Ananya", "Bharat", "Charan", "Divya", "Esha"],
    "region": ["South", "North", "South", "West", "East"]
})

orders = pd.DataFrame({
    "order_id": [1, 2, 3, 4, 5, 6],
    "customer_id": [101, 102, 103, 106, 101, 105],
    "product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Keyboard", "Laptop"],
    "sales": [55000, 1200, 2500, 15000, 2700, 60000]
})

display(customers)
display(orders)
```

```python
# Inner Join
inner_merge = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="inner"
)

print("INNER JOIN")
display(inner_merge)


# Left Join
left_merge = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="left"
)

print("LEFT JOIN")
display(left_merge)


# Right Join
right_merge = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="right"
)

print("RIGHT JOIN")
display(right_merge)


# Outer Join
outer_merge = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="outer",
    indicator=True
)

print("OUTER JOIN")
display(outer_merge)
```

```python
sales_data = pd.DataFrame({
    "region": [
        "South", "South", "North", "North", "West",
        "West", "East", "East", "South", "North"
    ],
    "product": [
        "Laptop", "Mouse", "Laptop", "Keyboard", "Laptop",
        "Mouse", "Keyboard", "Laptop", "Keyboard", "Mouse"
    ],
    "sales": [
        55000, 1200, 65000, 2500, 48000,
        1800, 3000, 52000, 2700, 2200
    ]
})

display(sales_data)
```

```python
sales_pivot = pd.pivot_table(
    sales_data,
    index="region",
    columns="product",
    values="sales",
    aggfunc="sum",
    fill_value=0
)

print("SALES BY REGION AND PRODUCT")

display(sales_pivot)
```

```python
region_summary = pd.pivot_table(
    sales_data,
    index="region",
    values="sales",
    aggfunc=["sum", "mean", "count"]
)

print("REGION SUMMARY")

display(region_summary)
```

```python
region_summary = pd.pivot_table(
    sales_data,
    index="region",
    values="sales",
    aggfunc=["sum", "mean", "count"]
)

print("REGION SUMMARY")

display(region_summary)
```

```python
wide_data = pd.DataFrame({
    "region": ["South", "North", "West"],
    "Laptop": [55000, 65000, 48000],
    "Mouse": [1200, 2200, 1800],
    "Keyboard": [2700, 2500, 3000]
})

print("WIDE DATA")

display(wide_data)


long_data = pd.melt(
    wide_data,
    id_vars=["region"],
    var_name="product",
    value_name="sales"
)

print("LONG DATA")

display(long_data)
```

```python
product_summary = (
    sales_data
    .groupby("product", as_index=False)
    .agg(
        total_sales=("sales", "sum"),
        average_sales=("sales", "mean"),
        number_of_orders=("sales", "count")
    )
    .sort_values("total_sales", ascending=False)
)

display(product_summary)

highest_product = product_summary.iloc[0]["product"]
highest_sales = product_summary.iloc[0]["total_sales"]

print("Product with highest sales:", highest_product)
print("Total sales: ₹{:,.2f}".format(highest_sales))
```
