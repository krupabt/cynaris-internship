```python
import numpy as np
import time
```

```python
# 1D Array
array_1d = np.array([10, 20, 30, 40, 50])

# 2D Array
array_2d = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

# 3D Array
array_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("1D Array:")
print(array_1d)

print("\n2D Array:")
print(array_2d)

print("\n3D Array:")
print(array_3d)

print("\nShapes:")
print("1D:", array_1d.shape)
print("2D:", array_2d.shape)
print("3D:", array_3d.shape)
```

```python
sales = np.array([100, 200, 300, 400, 500])

print("Original:", sales)
print("Add 50:", sales + 50)
print("Multiply by 2:", sales * 2)
print("Total:", np.sum(sales))
print("Mean:", np.mean(sales))
```

```python
data = np.array([
    [100, 200, 300],
    [400, 500, 600],
    [700, 800, 900]
])

print("Data:")
print(data)

print("\nSum across rows (axis=1):")
print(np.sum(data, axis=1))

print("\nSum across columns (axis=0):")
print(np.sum(data, axis=0))

print("\nMean across rows:")
print(np.mean(data, axis=1))

print("\nMean across columns:")
print(np.mean(data, axis=0))
```

```python
sales = np.array([12000, 25000, 18000, 45000, 30000])

selected = sales[[0, 2, 4]]

print("Original sales:")
print(sales)

print("\nSelected values:")
print(selected)
```

```python
sales = np.array([12000, 25000, 18000, 45000, 30000, 15000])

print("Mean:", np.mean(sales))
print("Standard Deviation:", np.std(sales))
print("25th Percentile:", np.percentile(sales, 25))
print("50th Percentile:", np.percentile(sales, 50))
print("75th Percentile:", np.percentile(sales, 75))
```

```python
advertising = np.array([10, 20, 30, 40, 50])
sales = np.array([100, 180, 250, 350, 450])

correlation_matrix = np.corrcoef(advertising, sales)

print("Correlation Matrix:")
print(correlation_matrix)

print("\nCorrelation coefficient:")
print(correlation_matrix[0, 1])
```

```python
# Create 1 million values
n = 1_000_000

numbers = np.arange(n)
python_list = list(range(n))

# Python loop
start = time.time()

python_result = []

for number in python_list:
    python_result.append(number * 2)

python_time = time.time() - start

# NumPy vectorised operation
start = time.time()

numpy_result = numbers * 2

numpy_time = time.time() - start

print("Python loop time:", python_time, "seconds")
print("NumPy vectorised time:", numpy_time, "seconds")

print("\nNumPy is approximately",
      round(python_time / numpy_time, 2),
      "times faster.")
```
